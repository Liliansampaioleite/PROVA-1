Crie um programa que deve validar a entrda de pessoas ,em uma sala de cinema, com base na idade e no valor disponivel para a compra do ingresso
Perguntar ao usuario o valor total que ele tem na carteira ,Se a idade for maior ou igual a 16 anos, verificar o dinheiro disponivel

```python
idade = int(input("Digite sua idade "))
valor_carteira = float(input("Digite o valor que está em sua carteira: "))

if idade < 16:
  print("Entrada permitida apenas com acompanhante ")
elif idade >= 16:
  print("verificar o dinheiro")
  if valor_carteira >= 30:
    print("Ingresso comprado, Bom filme ")
  else:
    print("Saldo insuficiente para comprar o ingresso")
else:
  print("Entrada proibida")

print("Foi bom te atender!!!")
```
